```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/processed/cleaned_data.csv")
df.head()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(df["SalePrice"], kde=True)
plt.title("Target Distribution")
plt.show()
```

```python
plt.figure(figsize=(12, 8))

# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation
correlation = numeric_df.corr()

# Display correlation values
display(correlation)

# Plot heatmap
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="OverallQual",
    y="SalePrice",
    data=df
)

plt.title("Overall Quantity Vs Sale Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="GrLivArea",
    y="SalePrice",
    data=df
)

plt.title("Living Area Vs Sales Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="YearBuilt",
    y="SalePrice",
    data=df
)

plt.title("Living Area Vs Sales Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x="GarageCars",
    y="SalePrice",
    data=df
)

plt.title("Garage Capacity vs Sales Price")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    x="Neighborhood",
    y="SalePrice",
    data=df
)

plt.xticks(rotation=90)
plt.show()
```
